###Get_Request
 Demonstrates a simple GET request using the `requests` library.Fetches data from a public API and prints the JSON response.Using the [jsonplaceholder.typicode.com
 ](https://jsonplaceholder.typicode.com)

````python
def get_example(id):
    try:
        url = f'https://jsonplaceholder.typicode.com/posts/{id}'
        response = requests.get(url, timeout=5)  # Set a timeout of 5 seconds
        response.raise_for_status()  # Raise HTTPError for bad responses (4xx or 5xx)
        data = response.json()
        return data
    except requests.exceptions.ConnectionError as e:
        print(f"Connection error: {e}")
        return None  # or handle the error differently
    except requests.exceptions.Timeout as e:
        print(f"Timeout error: {e}")
        return None  # or handle the error differently
    except requests.exceptions.HTTPError as e:
        print(f"HTTP error: {e}")
        return None  # or handle the error differently
    except requests.exceptions.RequestException as e:  # Catch-all for other requests errors
        print(f"General requests error: {e}")
        return None  # or handle the error differently
````

###Post Example
Demonstrates a simple POST request using the `requests` library.Sends JSON data to a public API and prints the response.

````python
def post_example():
    try:
        url = 'https://jsonplaceholder.typicode.com/posts'
        data = {
            'title': 'foo',
            'body': 'bar',
            'userId': 1
        }
        response = requests.post(url, json=data, timeout=5)  # Set a timeout of 5 seconds
        response.raise_for_status()  # Raise HTTPError for bad responses (4xx or 5xx)
        if response.status_code == 201:
            print("Post created successfully!")
            print(response.json())
        else:
            print(f"Failed to create post. Status code: {response.status_code}")
    except requests.exceptions.ConnectionError as e:
        print(f"Connection error: {e}")
    except requests.exceptions.Timeout as e:
        print(f"Timeout error: {e}")
    except requests.exceptions.HTTPError as e:
        print(f"HTTP error: {e}")
    except requests.exceptions.RequestException as e:  # Catch-all for other requests errors
        print(f"General requests error: {e}")
````

##Advance Request
custom headers and user agents play a significant role in HTTP requests. They allow you to send additional metadata or simulate specific client behavior when making requests to a server.


---


**Custom Headers:** Custom headers provide additional context or information to the server. They can be used for:

Authentication (Authorization header with tokens or API keys)
Content negotiation (e.g., Content-Type or Accept headers)
Caching control (Cache-Control, If-Modified-Since)
Sending API-specific data (e.g., X-API-Key)

---
**User-Agent Header:**
The User-Agent header identifies the type of client making the request (browser, app, bot, etc.). Servers can use this information to:

Serve different content or layouts for specific browsers or devices.
Restrict or allow access to certain bots (e.g., search engine crawlers).
Log analytics data about client usage.


A User-Agent string example for Google Chrome on Windows:


```
Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/117.0.0.0 Safari/537.36

```





###Get with headers
Demonstrates a GET request with custom headers and a user agent.Fetches data from a public API and prints the JSON response.

````python
def get_with_headers(id):
  url = f'https://jsonplaceholder.typicode.com/posts/{id}'
  headers = {
      'User-Agent': 'My User Agent 1.0',
      'Accept': 'application/json'
  }

  try:
      response = requests.get(url, headers=headers)
      response.raise_for_status()  # Raises an HTTPError for bad responses
      print("GET request with custom headers successful!")
      print(response.json())
      #print(response.text) i can use it for text format data
  except requests.exceptions.HTTPError as errh:
      print(f"HTTP Error: {errh}")
  except requests.exceptions.ConnectionError as errc:
      print(f"Connection Error: {errc}")
  except requests.exceptions.Timeout as errt:
      print(f"Timeout Error: {errt}")
  except requests.exceptions.RequestException as err:
      print(f"Error: {err}")
  except Exception as e:
      print(f"Unknown Error: {e}")
````

###Post with authentication
Demonstrates a POST request using basic authentication.Sends JSON data to a public API and prints the response.

````python
def post_with_authentication():
  url = 'https://jsonplaceholder.typicode.com/posts'
  data = {
      'title': 'foo',
      'body': 'bar',
      'userId': 1
  }
  headers = {
      'User-Agent': 'My User Agent 1.0'
  }
  auth = ('user', 'pass')  # Replace with actual username and password

  try:
      response = requests.post(url, json=data, headers=headers, auth=auth)
      response.raise_for_status()
      print("POST request with authentication successful!")
      print(response.json())
  except requests.exceptions.HTTPError as errh:
      print(f"HTTP Error: {errh}")
  except requests.exceptions.ConnectionError as errc:
      print(f"Connection Error: {errc}")
  except requests.exceptions.Timeout as errt:
      print(f"Timeout Error: {errt}")
  except requests.exceptions.RequestException as err:
      print(f"Error: {err}")
  except Exception as e:
      print(f"Unknown Error: {e}")
````

###Main Function
Main function to execute the examples.


````python
def main():
  print("Executiong Simple Request:")
  print("Executing Get Example for multiple request:")
  Data = []
  for i in range(1, 11):
      try:
          val = get_example(i)
          #Only append if val is not None
          if val is not None:
              Data.append(val)
      except requests.exceptions.RequestException:
          # Silently ignore errors and continue to the next iteration
          continue
  for index, item in enumerate(Data):
      print(f"post: {index}, Title: {item['title']}")
  print()
  post_example()
  print(get_example(101))
  print()
  print("Executing Advance Request")
  print("Executing GET request with custom headers...")
  get_with_headers(1)
  print("\nExecuting POST request with authentication...")
  post_with_authentication()
if __name__ == "__main__":
    main()
````

##Rendering Javscript
 Demonstrates how to render JavaScript using the `requests-html` library.
    This function fetches the page content after JavaScript has been executed.

    Parameters:
    url : str
        The URL of the website to scrape.

    Returns:
    None

````python
def render_javascript(url):
  session = HTMLSession()
  try:
    response = session.get(url)
    response.html.render() #this wiill download chromium if not found.
    print("Rendered web page:", response.html.html)
  except requests.exceptions.RequestException as e:
    print(f"Error rendering JavaScript: {e}")
  finally:
    session.close()
````

##Extract Information
Extracts and prints specific information from a webpage using CSS selectors.

    Parameters:
    url : str
        The URL of the website to scrape.

    Returns:
    None

````python
def extract_information(url):
  session = HTMLSession()
  try:
    response = session.get(url)
    #response.html.render() #this wiill download chromium if not found.
    #finding header
    title_tag = response.html.find('h1')
    print(len(title_tag),"Title tags found:")
    print("Title: ",title_tag[0].text)
    #Data time extraction
    date_element = response.html.find('time',first=True)#first occurs time will be render
    datetime = date_element.attrs.get('datetime')#.attrs die attribute access kore.
    print("Date:", datetime)
    print("Date: ",date_element.text)
    #Advance usages of extraction
    #Xpath of style can be used for extracting result
    temp = response.html.xpath('//*[@id="container"]/div/div[2]/div/div[1]/div[1]/div[1]/div[2]/div[2]/div[1]/time',first=True)
    print("date time again re extraced by xpath: ", temp)
    DateTime = temp.attrs.get('datetime')
    print("Date:", DateTime)

    print()

    #Exaple: Extracting all links
    links = response.html.find('a')
    for link in links:
      print("Link Text: ",link.text,"Link href: ",link.attrs.get('href'))
  except Exception as e:
    print(f"Error extracting information: {e}")
  finally:
    session.close()
````
